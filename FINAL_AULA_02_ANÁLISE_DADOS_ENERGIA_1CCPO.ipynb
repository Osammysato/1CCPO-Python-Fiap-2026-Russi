GERE UM HISTOGRAMA APENAS DA COLUNA Potencia_Ativa e me envie um relatório do motivo de existirem somente 372 de 160K consumindo acima de P60.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(dados['Potencia_Ativa'], kde=True, bins=50)
plt.axvline(x=P60, color='r', linestyle='--', label=f'Threshold (P60 = {P60:.2f} kW)')
plt.title('Distribution of Potencia_Ativa')
plt.xlabel('Potencia_Ativa (kW)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

### Análise do Limiar P60 e Consumo de Energia

O histograma da coluna 'Potencia_Ativa' (Potência Ativa) mostra a distribuição do consumo de energia no seu dataset. Podemos observar que a maioria dos registros de consumo de energia se concentra em valores baixos.

O P60 foi calculado como 60% da demanda máxima (`PMAX`), resultando em um valor de **6.17 kW**. A linha vertical vermelha no histograma indica este limiar. É evidente que uma porção muito pequena da distribuição de 'Potencia_Ativa' ultrapassa este valor.

Existem algumas razões prováveis para o fato de apenas 372 dos 163.943 registros de consumo estarem acima do limiar P60 (aproximadamente 0.23% do total):

1.  **Natureza dos Dados (Carga Base vs. Carga de Pico):** A maioria dos consumidores, ou a maior parte do tempo de consumo, opera em níveis de potência abaixo da capacidade máxima. Atingir 60% da potência máxima global (que é `PMAX`) é um evento relativamente raro, geralmente associado a picos de consumo ou a cargas de equipamentos específicos e de alta demanda.
2.  **Eficiência Energética:** Pode indicar que os equipamentos ou hábitos de consumo da maioria dos usuários são eficientes e não exigem constantemente altos níveis de potência.
3.  **Dimensionamento da Capacidade:** A capacidade máxima de 10.29 kW (`PMAX`) pode ser dimensionada para acomodar situações de carga muito elevadas, que não ocorrem com frequência na operação diária ou em todos os pontos de medição.
4.  **Amostragem ou Período de Tempo:** Se os dados representam um período específico ou são uma amostra, pode ser que os eventos de alto consumo sejam raros nesse período ou não tenham sido capturados em maior frequência.
5.  **Outliers:** Os valores que excedem o P60 podem ser considerados picos de demanda ou até mesmo *outliers* em relação ao comportamento de consumo médio, indicando momentos de uso intensivo de energia.

Em resumo, a pequena quantidade de registros acima de P60 sugere que a operação normal e a maioria dos padrões de consumo no dataset estão bem abaixo dos picos de demanda mais elevados.
